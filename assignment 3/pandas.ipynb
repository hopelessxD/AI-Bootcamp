```python
# load dataset and define student_df
import pandas as pd
import seaborn as sns

url = "https://raw.githubusercontent.com/GokuMohandas/Made-With-ML/main/datasets/titanic.csv"
try:
    student_df = pd.read_csv(url, header=0)
except Exception as e:
    print(f"Failed to load from URL ({e}), falling back to seaborn 'titanic' dataset.")
    student_df = sns.load_dataset('titanic')

# normalize column names to avoid case/whitespace issues
student_df.columns = [str(c).strip().lower() for c in student_df.columns]

student_df.head()
```

```python
# Creating pandas Series examples with renamed variables
num_list = [1, 2, 3, 4, 5]
mySeriesA = pd.Series(num_list)
mySeriesA
```

```python
# Series with custom index
num_list = [10, 20, 30, 40, 50]
my_index = ['a','b','c','d','e']
mySeriesB = pd.Series(num_list, index=my_index)
print(mySeriesB)
type(mySeriesB)
```

```python
# Mixed-type Series with a custom name
mixed_data = [0.5, 'missing', 404]
idx = ['x','y','z']
mySeriesC = pd.Series(mixed_data, idx, name='MyErrors')
mySeriesC.index.name = 'Index'
print(mySeriesC)
```

```python
# Series from a dictionary
dict_sample = {'a': 100, 'b': 200, 'c': 300, 'd': 400, 'e': 'text'}
mySeriesD = pd.Series(dict_sample, name='Col1')
print(mySeriesD.index)
mySeriesD
```

```python
# Series from list of lists
list_of_rows = [[1,2,3],[4,5,6],[7,8,9]]
idx = [1,2,3]
mySeriesE = pd.Series(list_of_rows, idx)
print(mySeriesE[1])
print(type(mySeriesE[1]))
mySeriesE
```

```python
# Create DataFrame from Series
data_map = {'ColA': mySeriesD, 'ColB': mySeriesB}
my_dataframe = pd.DataFrame(data_map)
my_dataframe
```

```python
# DataFrame from dictionaries
dict1 = {'a':1, 'b':2, 'c':3, 'd':4}
dict2 = {'a':'m', 'b':'n', 'c':'o', 'd':'p', 'e':'q'}
combined = {'Left': dict1, 'Right': dict2}
combined_df = pd.DataFrame(combined)
combined_df
```

```python
# DataFrame from list of dicts
dict_list = [
	{'x1':1, 'x2':2, 'x3':3, 'x4':4},
	{'x1':5, 'x2':6, 'x3':7},
	{'x1':8, 'x2':9, 'x3':10, 'x4':11, 'x5':12}
]
pd.DataFrame.from_dict(dict_list, orient='columns')
```

```python
# Data cleaning example (using renamed vars)
sample_copy = student_df.copy()
for i in range(0, 10000, 100):
		if i in sample_copy.index:
				sample_copy.loc[i, 'Hours Studied'] = np.nan
sample_copy.isnull().sum()
```

```python
# Drop nulls and duplicates
cleaned = sample_copy.dropna(axis=0)
cleaned = cleaned.drop_duplicates()
cleaned.shape
```
